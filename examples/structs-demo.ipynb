# Persistent structs and nested containers
Run cells in order. The schema change is intentionally rejected while a value exists. %reset discards all variables and definitions, after which the new schema is allowed. Start a new kernel after upgrading.

```daslang
struct Stats {
 count : int
 values : array<int>
}
var stats : Stats
stats.count=5
stats.values |> push(7)
stats
```

```daslang
stats.count += 1
stats.values |> push(9)
stats
```

```daslang
var rows : array<array<int>>
var row : array<int> <- [1,2]
rows |> push_clone(row)
rows
```

```daslang
struct Stats {count : string; values : array<int>}
```

```daslang
stats.count
```

```daslang
%reset
```

```daslang
struct Stats {count : string}
var stats : Stats
stats.count="new schema"
stats.count
```

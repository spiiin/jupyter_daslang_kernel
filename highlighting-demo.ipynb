# Подсветка daslang
Цвета следуют теме JupyterLab. Функции, типы, числа, строки и комментарии подсвечиваются без выполнения кода. Выберите ядро daslang.

```daslang
// daslang highlighting
/* Outer comment
   /* Nested comment */
*/
struct Stats {
    count : int
    values : array<int>
}
var stats : Stats
stats.count = 3
stats.values |> push(int(0x2a))
def twice(n : int) : int { return n * 2 }
print("Привет: {twice(stats.count)}\n")
stats
```

```daslang
"Привет {stats.count}, массив: {stats.values}"
```

```daslang
%eval twice(stats.count)
```

# Stateful daslang: mixed cells and inferred types
Run cells with Shift+Enter. Declarations, functions and statements can share a cell. Expected: x=15, y=20, values[0]=15, twice()=30 then 45. The intentional compile error preserves state.

```daslang
def calc_primes(n: int; var a:array<int>) {
    a |> clear
    a |> reserve(n)

    a |> push(2)
    if (n == 1) { return; }
    a |> push(3)

    var gen <- generator<int>() <| $() {
        var t = 6
        while (true) {
            yield t-1
            yield t+1
            t += 6
        }
        return false
    }

    var iterVal: int
    while (length(a) < n) {
        next(gen, iterVal)
        var simple = true
        for (prim in a) {
            if (iterVal % prim == 0) {
                simple = false
                break
            }
            if (prim * prim > iterVal) {
                break
            }
        }
        if (simple) {
            a |> push(iterVal)
        }
    }
}
```

```daslang
var a: array<int>
calc_primes(1000000, a)
print("{a[length(a)-1]}\n")
```

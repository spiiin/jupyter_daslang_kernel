# let, перегрузки и require
Выберите daslang и перезапустите ядро после обновления до 0.9. Все определения и импорты сохраняются между ячейками. Ошибка записи в let в конце примера ожидаема.

Импорт функций и констант из `.das`-модулей поддерживается. Используемые runtime-глобалы и lifecycle hooks импортируемых модулей пока отклоняются до исполнения.

```daslang
require math as m
require daslib/strings_boost as sb
```

```daslang
var calls = 0
def next() : int { calls += 1; return calls }
let saved = next()
let root = m::sqrt(9.0)
let words <- sb::split("a,b", ",")
```

```daslang
"saved={saved}, calls={calls}, root={root}, words={words}"
```

```daslang
//! Описывает целое число.
def describe(x : int) : string { return "int:{x}" }
//! Описывает строку.
def describe(x : string) : string { return "string:{x}" }
```

```daslang
describe(saved)
```

```daslang
describe("das")
```

Переопределяем только перегрузку с параметром int. Строковая перегрузка остаётся прежней. Shift+Tab у `describe(` показывает обе сигнатуры.

```daslang
def describe(value : int) : string { return "updated:{value * 10}" }
```

```daslang
describe(saved)
```

```daslang
describe("das")
```

```daslang
calls
```

Следующая ячейка должна выдать ошибку компилятора: saved — неизменяемая переменная. После ошибки запустите последнюю ячейку: saved по-прежнему равно 1.

```daslang
saved = 99
```

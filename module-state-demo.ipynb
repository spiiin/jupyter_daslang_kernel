# Состояние импортируемого модуля
Пример требует ядро daslang 0.10 и файл `examples/notebook_counter.das` из каталога проекта. После обновления перезапустите ядро.

`[init]` вызывается при первом импорте. Приватные глобалы, массив и таблица сохраняются между ячейками. Перекомпиляция не запускает init/finalize повторно. `%reset` завершает lifecycle и очищает состояние.

```daslang
require ./examples/notebook_counter.das as counter
```

```daslang
counter::next()
```

```daslang
counter::next()
```

```daslang
counter::state()
```

Следующая ячейка добавляет обычное состояние notebook. Импортируемый модуль сохраняет прежние значения.

```daslang
let notebook_value = 7
notebook_value
```

```daslang
counter::state()
```

Сброс должен напечатать один finalize. Затем повторный импорт запускает новый init.

```daslang
%reset
```

```daslang
require ./examples/notebook_counter.das as counter
counter::state()
```

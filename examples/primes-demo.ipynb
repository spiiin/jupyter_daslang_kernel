# Prime numbers in Gen2
The original function used control statements without parentheses and indentation blocks. This version uses Gen2 braces and a mutable generator. The last expression automatically shows the array. Restart the kernel after upgrading.

```daslang
def calc_primes(n : int; var a : array<int>) {
    a |> clear()
    if (n <= 0) {
        return
    }
    a |> reserve(n)
    a |> push(2)
    if (n == 1) {
        return
    }
    a |> push(3)

    var gen <- generator<int>() <| $() {
        var t = 6
        while (true) {
            yield t - 1
            yield t + 1
            t += 6
        }
        return false
    }

    var iterVal : int
    while (length(a) < n) {
        next(gen, iterVal)
        var simple = true
        for (prim in a) {
            if (iterVal % prim == 0) {
                simple = false
                break
            }
            if (prim * prim > iterVal) {
                break
            }
        }
        if (simple) {
            a |> push(iterVal)
        }
    }
}
```

```daslang
var primes : array<int>
calc_primes(10, primes)
primes
```

```daslang
calc_primes(1, primes)
primes
```

```daslang
calc_primes(0, primes)
primes
```

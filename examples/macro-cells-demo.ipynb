# Макросы daslang из отдельного файла

Первая ячейка сохраняет модуль рядом с notebook. Следующие ячейки подключают его и используют настоящий compile-time макрос.

```daslang
%%writefile notebook_macros.das
options gen2
module notebook_macros public
require daslib/ast
require daslib/ast_boost
require daslib/templates_boost
require daslib/macro_boost
require daslib/ast_verify
[call_macro(name="plus_one")]
class PlusOneMacro : AstCallMacro {
    def override visit(prog:ProgramPtr; mod:Module?; var expr:ExprCallMacro?):ExpressionPtr {
        macro_verify(length(expr.arguments)==1, prog, expr.at, "plus_one expects one argument")
        var result <- qmacro($e(clone_expression(expr.arguments[0])) + 1)
        verify_expression(result)
        return <- result
    }
}
```

```daslang
require ./notebook_macros.das
var value=40
plus_one(value)
```

Изменим развёртку макроса с `+1` на `+2`, затем выполним обычную ячейку. Значение `value` сохранится.

```daslang
%%writefile notebook_macros.das
options gen2
module notebook_macros public
require daslib/ast
require daslib/ast_boost
require daslib/templates_boost
require daslib/macro_boost
require daslib/ast_verify
[call_macro(name="plus_one")]
class PlusOneMacro : AstCallMacro {
    def override visit(prog:ProgramPtr; mod:Module?; var expr:ExprCallMacro?):ExpressionPtr {
        macro_verify(length(expr.arguments)==1, prog, expr.at, "plus_one expects one argument")
        var result <- qmacro($e(clone_expression(expr.arguments[0])) + 2)
        verify_expression(result)
        return <- result
    }
}
```

```daslang
plus_one(value)
```

```daslang
value
```

Макрос вычисляется во время компиляции. Состояние ячеек хранится отдельно. Изменения файла не откатываются при Interrupt или `%reset`.
